```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')
```

```python
df = pd.read_csv('ecommerce_customer_data_custom_ratios.csv.zip')
```

```python
print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst Five Rows:")
display(df.head())

print("\nDataset Information:")
df.info()
```

```python
print("\nStatistical Summary:")
display(df.describe(include='all'))

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
print("\nReturns Distribution:")
print(df['Returns'].value_counts())

print("\nChurn Distribution:")
print(df['Churn'].value_counts())

print("\nUnique Customers:", df['Customer ID'].nunique())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.countplot(data=df, x='Returns', ax=axes[0])
axes[0].set_title('Product Returns Distribution')

sns.countplot(data=df, x='Churn', ax=axes[1])
axes[1].set_title('Customer Churn Distribution')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='Product Category')
plt.title('Product Category Distribution')
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='Product Category')
plt.title('Product Category Distribution')
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.histplot(df['Total Purchase Amount'], bins=30, kde=True)
plt.title('Purchase Amount Distribution')
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x='Product Category',
            y='Returns', errorbar=None)
plt.title('Return Rate by Product Category')
plt.ylabel('Average Return Rate')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Gender', y='Churn', errorbar=None)
plt.title('Churn Rate by Gender')
plt.ylabel('Average Churn Rate')
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.histplot(df['Customer Age'], bins=20, kde=True)
plt.title('Customer Age Distribution')
plt.show()
```

```python
churn_check = df.groupby('Customer ID')['Churn'].nunique()

print("\nCustomers with consistent churn labels:",
      (churn_check == 1).sum())

print("Customers with inconsistent churn labels:",
      (churn_check > 1).sum())
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(11, 7))
sns.heatmap(numeric_df.corr(), annot=True,
            cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()

print("\nData Exploration Completed!")
```
